TASK 1:Use AI to generate test cases for a function is_prime(n) and then implement the function.
Requirements:
• Only integers > 1 can be prime.
•Check edge cases: 0, 1, 2, negative numbers, and large prime

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    sqrt_n = int(math.sqrt(n)) + 1
    for i in range(3, sqrt_n, 2):
        if n % i == 0:
            return False
test_cases = [
    (-10, False),
    (-1, False),
    (0, False),
    (1, False),
    (2, True),
    (3, True),
    (4, False),
    (17, True),
    (100, False),
    (101, True),
    (7919, True),
    (9999, False),
]

for i, (n, expected) in enumerate(test_cases, 1):
    result = is_prime(n)
    print(f"Test Case {i}: is_prime({n}) == {expected} --> {'PASS' if result == expected else 'FAIL'}")
```

TASK 2:Ask AI to generate test cases for celsius_to_fahrenheit(c) and fahrenheit_to_celsius(f).
Requirements
• Validate known pairs: 0°C = 32°F, 100°C = 212°F.
• Include decimals and invalid inputs like strings or None

```python
import pytest

# Define the functions here
def celsius_to_fahrenheit(c):
    if not isinstance(c, (int, float)):
        raise TypeError("Input must be a number.")
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    if not isinstance(f, (int, float)):
        raise TypeError("Input must be a number.")
    return (f - 32) * 5/9

# Known value pairs
def test_celsius_to_fahrenheit_known():
    assert celsius_to_fahrenheit(0) == 32
    assert celsius_to_fahrenheit(100) == 212

def test_fahrenheit_to_celsius_known():
    assert fahrenheit_to_celsius(32) == 0
    assert fahrenheit_to_celsius(212) == 100

# Decimal inputs
def test_celsius_to_fahrenheit_decimals():
    assert round(celsius_to_fahrenheit(37.5), 1) == 99.5
    assert round(celsius_to_fahrenheit(-40.0), 1) == -40.0

def test_fahrenheit_to_celsius_decimals():
    assert round(fahrenheit_to_celsius(98.6), 1) == 37.0
    assert round(fahrenheit_to_celsius(-40.0), 1) == -40.0

# Invalid inputs
@pytest.mark.parametrize("invalid_input", ["abc", None, [10], {"temp": 30}])
def test_celsius_to_fahrenheit_invalid(invalid_input):
    with pytest.raises(TypeError):
        celsius_to_fahrenheit(invalid_input)

@pytest.mark.parametrize("invalid_input", ["xyz", None, [32], {"temp": 100}])
def test_fahrenheit_to_celsius_invalid(invalid_input):
    with pytest.raises(TypeError):
        fahrenheit_to_celsius(invalid_input)

print("Celsius to Fahrenheit:")
print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(100))
```

TASK3:write test cases for a function count_words(text) that returns the number of
words in a sentence.
Requirement
Handle normal text, multiple spaces, punctuation, and empty strings

```python
import pytest

# Define the function here
def count_words(text):
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
    words = text.split()
    return len(words)

# --- Normal sentences ---
def test_count_words_normal():
    assert count_words("Hello world") == 2
    assert count_words("This is a test sentence.") == 5

# --- Multiple spaces ---
def test_count_words_with_multiple_spaces():
    assert count_words("   Hello    world   again ") == 3
    assert count_words("Word     spacing     test") == 3

# --- Punctuation included ---
def test_count_words_with_punctuation():
    assert count_words("Wait... what?!") == 2
    assert count_words("Well, this is odd.") == 4

# --- Empty string or spaces only ---
def test_count_words_empty_input():
    assert count_words("") == 0
    assert count_words("   ") == 0

# --- Non-string input (invalid) ---
@pytest.mark.parametrize("invalid_input", [None, 123, ["word"], {"text": "hello"}])
def test_count_words_invalid_input(invalid_input):
    with pytest.raises(TypeError):
        count_words(invalid_input)
print("Normal sentences:")
print(count_words("Hello world"))
print(count_words("This is a test sentence."))
```

TASK 4:Generate test cases for a BankAccount class with: Methods: deposit(amount) withdraw(amount) check_balance() Requirements: • Negative deposits/withdrawals should raise an error. • Cannot withdraw more than balance

```python
import pytest

class BankAccount:
    def __init__(self, initial_balance=0):
        if not isinstance(initial_balance, (int, float)) or initial_balance < 0:
            raise ValueError("Initial balance must be a non-negative number")
        self.balance = initial_balance

    def deposit(self, amount):
        if not isinstance(amount, (int, float)) or amount <= 0:
            raise ValueError("Deposit amount must be a positive number")
        self.balance += amount

    def withdraw(self, amount):
        if not isinstance(amount, (int, float)) or amount <= 0:
            raise ValueError("Withdrawal amount must be a positive number")
        if amount > self.balance:
            raise ValueError("Cannot withdraw more than current balance")
        self.balance -= amount

    def check_balance(self):
        return self.balance

# --- Fixture to create a new bank account ---
@pytest.fixture
def account():
    return BankAccount()

# --- Initial balance check ---
def test_initial_balance(account):
    assert account.check_balance() == 0

# --- Valid deposit ---
def test_deposit(account):
    account.deposit(100)
    assert account.check_balance() == 100

# --- Valid withdrawal ---
def test_withdraw(account):
    account.deposit(200)
    account.withdraw(50)
    assert account.check_balance() == 150

# --- Negative deposit ---
def test_negative_deposit(account):
    with pytest.raises(ValueError, match="positive"):
        account.deposit(-50)

# --- Negative withdrawal ---
def test_negative_withdrawal(account):
    with pytest.raises(ValueError, match="positive"):
        account.withdraw(-100)

# --- Withdrawal more than balance ---
def test_withdraw_more_than_balance(account):
    account.deposit(100)
    with pytest.raises(ValueError, match="Insufficient"):
        account.withdraw(150)

# --- Zero deposit or withdrawal ---
def test_zero_deposit_withdraw(account):
    with pytest.raises(ValueError):
        account.deposit(0)
    with pytest.raises(ValueError):
        account.withdraw(0)

# --- Multiple transactions ---
def test_multiple_transactions(account):
    account.deposit(500)
    account.withdraw(200)
    account.deposit(300)
    account.withdraw(100)
    assert account.check_balance() == 500
print("Multiple transactions:")
print(account)
```

TASK 5:Generate test cases for a BankAccount class with: Methods: deposit(amount) withdraw(amount) check_balance() Requirements: • Negative deposits/withdrawals should raise an error. • Cannot withdraw more than balance

```python
import pytest

def is_number_palindrome(num):
    """Checks if an integer reads the same backward."""
    if not isinstance(num, int):
        return False  # Handle non-integer input
    if num < 0:
        return False  # Handle negative numbers

    # Convert the integer to a string to easily check for palindrome
    num_str = str(num)
    return num_str == num_str[::-1]

# Test cases for is_number_palindrome function
class TestIsNumberPalindrome:

    def test_examples(self):
        assert is_number_palindrome(121) is True
        assert is_number_palindrome(123) is False

    def test_single_digit(self):
        assert is_number_palindrome(0) is True
        assert is_number_palindrome(5) is True

    def test_negative_numbers(self):
        assert is_number_palindrome(-121) is False
        assert is_number_palindrome(-1) is False

    def test_multi_digit_palindromes(self):
        assert is_number_palindrome(1221) is True
        assert is_number_palindrome(98789) is True

    def test_multi_digit_non_palindromes(self):
        assert is_number_palindrome(1234) is False
        assert is_number_palindrome(123210) is False

    def test_non_integer_input(self):
        assert is_number_palindrome(121.0) is False
        assert is_number_palindrome("121") is False
        assert is_number_palindrome(None) is False
print(f"Test case {is_number_palindrome}:passed")
```
